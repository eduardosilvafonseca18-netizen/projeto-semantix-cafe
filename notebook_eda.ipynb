# EDA — café arábica, Espera Feliz e estados produtores
Fonte: IBGE/PAM/SIDRA 1613 (2014–2024). As consultas e metodologia estão no README.


```python
import pandas as pd
from pathlib import Path
df = pd.read_csv(Path("dados/cafe_arabica_2014_2024.csv"))
df.head()
```

## Qualidade e pré-processamento
A coleta normaliza marcadores de ausência e une as variáveis por localidade e ano.


```python
print(df.shape)
print(df.isna().sum())
print("Duplicatas:", df.duplicated(["escala", "codigo_ibge", "ano"]).sum())
assert (df["area_colhida_ha"] > 0).all()
assert (df["producao_t"] >= 0).all()
```

## Distribuição e tendências


```python
print(df.groupby("localidade")["sacas_ha"].describe().round(2))
print(df.pivot(index="ano", columns="localidade", values="sacas_ha").round(2))
```

```python
import matplotlib.pyplot as plt
ax=df.pivot(index="ano", columns="localidade", values="sacas_ha").plot(figsize=(10,5), marker="o", title="Produtividade do café arábica")
ax.set_ylabel("Sacas de 60 kg por hectare")
plt.show()
```

## Relações e cautelas
A correlação é descritiva. Produção, área e produtividade têm relação matemática; valor da produção e valor implícito também. Evite interpretar esses coeficientes como efeitos causais.


```python
for nome, fatia in df.groupby("localidade"):
    print(nome)
    print(fatia[["area_colhida_ha", "producao_t", "sacas_ha", "valor_implicito_reais_saca"]].corr().round(2))
```
